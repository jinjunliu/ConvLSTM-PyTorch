```python
import matplotlib.pyplot as plt
import numpy as np
    
# Sample latitude and longitude data
lat = np.linspace(-90, 91, 90)
lon = np.linspace(-180, 181, 180)

# Create a meshgrid from lat/lon
lon_grid, lat_grid = np.meshgrid(lon, lat)

print(lon_grid.shape, lat_grid.shape)

# Sample data values (e.g., a function of lat and lon)
data = np.sin(np.deg2rad(lat_grid)) * np.cos(np.deg2rad(lon_grid))

print(data.shape)

plt.figure(figsize=(10, 6))

# Filled contour plot
contour = plt.contourf(lon, lat, data, cmap='viridis')  # or other colormap

# Add colorbar
plt.colorbar(contour)

# Add labels and title
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("Filled Contour Plot of Sample Data")

# Show the plot
plt.show()
```
